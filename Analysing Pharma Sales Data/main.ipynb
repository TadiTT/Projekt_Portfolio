```python
from turtledemo.clock import datum

### Imports die man für das Projekt brauch
import matplotlib.pyplot as plt
import pandas as pd
```

```python
###Aufgabe 1 - Was sind die Gesamtumsatzmengen für jede Arzneimittelkategorie (ATC-Code)?###
df = pd.read_csv("salesdaily.csv")

###Vorbereitung###
atc_spalten = ['M01AB', 'M01AE', 'N02BA', 'N02BE', 'N05B', 'N05C', 'R03', 'R06']

gesamtumstz = df[atc_spalten].sum().sort_values(ascending=False)

print("=== Gesamtumsatz je Arzneimittelkategorie ===")
print(gesamtumstz.sort_values(ascending=False).round(2))

###Visualisierung###
plt.barh(atc_spalten, gesamtumstz.values)

plt.title("Gesamtumsatzmengen für jede Arzneimittelkategorie (ATC-Code)", fontsize=20)
plt.xlabel("Gesamtumsatz", fontsize=20)
plt.xlabel("Gesamtumsatz in Euro", family = "Arial", fontsize = 15)
plt.ylabel("Kategorien", family="Arial", fontsize = 15)
plt.grid(axis = "y", linestyle = "--", linewidth = 0.5)
plt.grid(axis = "x", linestyle = "--", linewidth = 0.5)
plt.show()
```

```python
###Aufgabe 2 - Welche einzelnen Arzneimittelmarken haben den höchsten Gesamtumsatz?###

###Vorbereitung###
atc_spalten = ['M01AB', 'M01AE', 'N02BA', 'N02BE', 'N05B', 'N05C', 'R03', 'R06']

gesamtumstz = df[atc_spalten].sum().sort_values(ascending=False).head(3)

print("=== Top 3 des Gesamtumsatzes je Arzneimittelkategorie ===")
print(gesamtumstz.sort_values(ascending=False).round(2))

###Visualisierung###
plt.barh(gesamtumstz.index, gesamtumstz.values)

plt.title("Gesamtumsatzmengen für jede Arzneimittelkategorie (ATC-Code)", fontsize=20)
plt.xlabel("Gesamtumsatz", fontsize=20)
plt.xlabel("Gesamtumsatz in Euro", family = "Arial", fontsize = 15)
plt.ylabel("Kategorien", family="Arial", fontsize = 15)
plt.grid(axis = "y", linestyle = "--", linewidth = 0.5)
plt.grid(axis = "x", linestyle = "--", linewidth = 0.5)
plt.show()
```

```python
###Aufgabe 3 - Welche drei Medikamente haben im Januar 2015, Juli 2016, September 2017 den höchsten Umsatz###

###Vorbereitung###
###Januar 2015###
januar_2015 = df[(df['Year'] == 2015) & (df['Month'] == 1)]
umsatz_jan = januar_2015[atc_spalten].sum()
top3_jan = umsatz_jan.nlargest(3)
print("=== 2015 Januar ===")
print(top3_jan)
###Juli 2016###
juli_2016 = df[(df['Year'] == 2016) & (df['Month'] == 7)]
umsatz_juli = juli_2016[atc_spalten].sum()
top3_juli = umsatz_juli.nlargest(3)
print("=== 2016 Juli ===")
print(top3_juli)
###September 2017###
september_2017 = df[(df['Year'] == 2017) & (df['Month'] == 9)]
umsatz_september = september_2017[atc_spalten].sum()
top3_september = umsatz_september.nlargest(3)
print("=== 2017 September ===")
print(top3_september)

###Visualisierung###
fig, axs = plt.subplots(3, figsize=(6, 10))
fig.suptitle("Top 3 Medikamente je Zeitraum", fontsize=20)

# Januar 2015
axs[0].bar(top3_jan.index, top3_jan.values)
axs[0].set_title('Januar 2015')
axs[0].set_xlabel('Medikament')
axs[0].set_ylabel('Umsatz')

# Juli 2016
axs[1].bar(top3_juli.index, top3_juli.values)
axs[1].set_title('Juli 2016')
axs[1].set_xlabel('Medikament')
axs[1].set_ylabel('Umsatz')

# September 2017
axs[2].bar(top3_september.index, top3_september.values)
axs[2].set_title('September 2017')
axs[2].set_xlabel('Medikament')
axs[2].set_ylabel('Umsatz')

plt.tight_layout()
plt.show()
```

```python
###Aufgabe 4 - Welches Medikament wurde 2017 am häufigsten verkauft?###

import pandas as pd
import matplotlib.pyplot as plt

atc_spalten = ['M01AB', 'M01AE', 'N02BA', 'N02BE', 'N05B', 'N05C', 'R03', 'R06']

# Zelle 2 — Aufgabe 3
df = pd.read_csv("salesdaily.csv")

###Vorbereitung###
top_2017 = df[(df['Year'] == 2017)]
umsatz_2017 = top_2017[atc_spalten].sum()
top1_2017index = umsatz_2017.idxmax()

print("=== Das meist verkaufteste Medikament 2017 ===")
print(f"Das Medikament: {top1_2017index}")


```

```python
###Aufgabe 5 - Welche Arzneimittelkategorie hat den höchsten durchschnittlichen täglichen Umsatz?###

###Vorbreitung###
atc_spalten = ['M01AB', 'M01AE', 'N02BA', 'N02BE', 'N05B', 'N05C', 'R03', 'R06']
durchschnitt = df[atc_spalten].mean()
print(durchschnitt.sort_values(ascending=False))

###Visualisierung###
plt.bar(durchschnitt.index, durchschnitt.values)
plt.title("Höchster durchschnittlicher Tagesumsatz")
plt.xlabel("Kategorie")
plt.ylabel("Ø Tagesumsatz")
plt.show()
```

```python
###Aufgabe 6 - Werden Atemwegsmedikamente (R03) in bestimmten Monaten mehr verkauft?###

###Vorbereitung###
r03_monatlich = df.groupby('Month')['R03'].sum()

print(r03_monatlich)

###Visualisierung###
plt.plot(r03_monatlich.index, r03_monatlich.values, marker='o')
plt.title("R03 Umsatz nach Monaten")
plt.xlabel("Monat")
plt.ylabel("Umsatz R03")
plt.xticks(range(1, 13))  # Monate 1-12 auf X-Achse
plt.grid(True)
plt.show()
```
