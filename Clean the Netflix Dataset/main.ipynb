```python
###Aufgabe 3 - Fix gemischte Spalten (z. B. Dauer, gespeichert als "90 min")###

import pandas as pd

df = pd.read_csv('netflix_titles.csv')

# Zahl aus dem String extrahieren
df['duration_int'] = df['duration'].str.extract(r'(\d+)').astype('Int64')

# Je nach Typ die richtige Spalte befüllen
df['duration_minutes'] = df['duration_int'].where(df['type'] == 'Movie')
df['duration_seasons'] = df['duration_int'].where(df['type'] == 'TV Show')

# Hilfsspalte aufräumen
df.drop(columns=['duration_int'], inplace=True)

print(df[['type', 'duration', 'duration_minutes', 'duration_seasons']])
```

```python
###Aufgabe 4 - Parse date columns into proper datetime objects###

import pandas as pd

df = pd.read_csv("netflix_titles.csv")

print(df['date_added'].head())

# Konvertierung zu "Jahr - Monat(int) - Tag"
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), format='%B %d, %Y')

print(df['date_added'])
```

```python
### Aufgabe 5 - Export the cleaned DataFrame to a new CSV file###

import pandas as pd

###Aufgabe 1 - Inspect the DataFrame with .info(), .describe(), and .head()###

df = pd.read_csv("netflix_titles.csv")

###.info()###
print("=== .info() ===")
print(df.info())
###.describe().###
print("=== .describe() ===")
print(df.describe())
###.head()###
print("=== .head() ===")
print(df.head())

###Aufgabe 2 - Spalte fehlende Werte identifizieren und handhaben###

# Identifizieren
df.isnull().sum()
df.isnull().sum() / len(df)

# Mit anderen Werten ersetzen
df['director'].fillna("Unbekannt", inplace=True)
df['cast'].fillna("Unbekannt", inplace=True)
df['country'].fillna("Unbekannt", inplace=True)


###Aufgabe 3 - Fix gemischte Spalten (z. B. Dauer, gespeichert als "90 min")###

# Zahl aus dem String extrahieren
df['duration_int'] = df['duration'].str.extract(r'(\d+)').astype('Int64')

# Je nach Typ die richtige Spalte befüllen
df['duration_minutes'] = df['duration_int'].where(df['type'] == 'Movie')
df['duration_seasons'] = df['duration_int'].where(df['type'] == 'TV Show')

# Hilfsspalte aufräumen
df.drop(columns=['duration_int'], inplace=True)

print(df[['type', 'duration', 'duration_minutes', 'duration_seasons']])

###Aufgabe 4 - Parse date columns into proper datetime objects###

print(df['date_added'].head())

# Konvertierung zu "Jahr - Monat(int) - Tag"
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), format='%B %d, %Y')

print(df['date_added'])


###Export###
df.to_csv('netflix_titles_cleaned.csv', index=False, encoding='utf-8-sig')

print("Fertig! Datei gespeichert als 'netflix_titles_cleaned.csv'")
print(f"   Zeilen: {len(df)} | Spalten: {len(df.columns)}")
```
